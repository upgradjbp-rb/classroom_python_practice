## ⚡ Module 1: Introduction to NumPy & List vs. Array Mechanics

### 1. Concept & Theory
NumPy (Numerical Python) provides high-performance multidimensional arrays (`ndarray`) and vectorized mathematical operations. Unlike native Python lists that store references to fragmented objects, NumPy allocates continuous blocks of homogeneous memory, eliminating loop interpretation overhead.

### 2. Syntax & Variations
```python
import numpy as np

# Vectorized array scaling vs list comprehension
arr = np.array([30, 32, 35, 28])
arr_res = arr + 1
list_res = [x + 1 for x in [30, 32, 35, 28]]
```

### 3. Points to Remember & Common Pitfalls
* `list * 2` repeats/concatenates the sequence (`[1, 2] * 2 -> [1, 2, 1, 2]`), whereas `arr * 2` multiplies each element by 2 mathematically (`[2, 4]`).
* Always import NumPy under the standard community alias `import numpy as np`.

## 🎯 Module 3: Array Indexing, Slicing, Fancy & Boolean Filtering

### 1. Concept & Theory
Array elements are accessed by integer coordinates:
* **Basic Indexing & Slicing**: `arr[start:stop:step]` or `matrix[row_slice, col_slice]` extracts contiguous subsets.
* **Negative Indexing**: Accesses elements relative to array endpoints (`arr[-1]`).
* **Fancy Indexing**: Passes an explicit list/array of integer indices (`arr[[0, 2, 4]]`) to retrieve arbitrary coordinates.
* **Boolean Indexing**: Evaluates conditional expressions to filter elements dynamically (`scores[scores >= 50]`).

### 2. Syntax & Variations
```python
arr[0:2, 1:3]                   # 2D submatrix
arr[::2]                        # Step slicing
arr[::-1]                       # Step reversal
arr[indices]                    # Fancy indexing
arr[(arr >= 50) & (arr <= 80)]  # Compound boolean masking
np.take(arr, [0, 2])            # Programmatic index extraction
```

## 📡 Module 4: Array Operations, Broadcasting, & Aggregations

### 1. Concept & Theory
Operations on NumPy arrays are performed element-wise. When dimensions differ, **broadcasting** dynamically stretches arrays of dimension 1 to match the larger array without making redundant data copies.

### 2. Broadcasting Rules & Reductions
* **Rule 1**: Align shape tuples from right to left.
* **Rule 2**: If ranks differ, prepend 1s to the smaller array.
* **Rule 3**: Any dimension of size 1 stretches to match the corresponding dimension size of the other array.
* **Rule 4**: If dimension sizes differ and neither is 1, a `ValueError` occurs.
* **Aggregations**: `axis=0` aggregates column-wise (down rows); `axis=1` aggregates row-wise (across columns).

## 🔍 Module 7: Memory Sharing, Views, Copies, & Mutation Diagnostics

### 1. Concept & Theory
Managing memory prevents silent data corruption:
* **View (Shallow Reference)**: A slice `arr[1:3]` shares the underlying memory buffer with the parent array. Changing the view alters the original array.
* **Copy (Deep Duplicate)**: `arr.copy()` allocates an isolated memory block. Mutating a copy preserves the source data.
* **Verification**: `np.shares_memory(a, b)` explicitly verifies whether two arrays share address space.

```
Real-life Analogy:
• View -> Looking at a shared office whiteboard. Marking it alters it for everyone.
• Copy -> Taking a photo of the whiteboard to annotate at home. The board stays unchanged.
```

## 🩹 Module 15: Missing Data & Sentinel Values (`np.nan` Semantics)

### 1. Concept & Theory
Missing values are represented as `np.nan` (Not a Number):
* `np.isnan(arr)`: Returns a boolean mask identifying NaN cells.
* `np.nan_to_num(arr, nan=val)`: Replaces missing values with a designated constant.
* `np.where(np.isnan(data), mean, data)`: Conditionally imputes missing values.
* `np.nanmean()`, `np.nanstd()`, `np.nansum()`: Aggregates data while ignoring missing entries.
* `arr[~np.isnan(arr)]`: Filters out missing entries.

````python
# [SOLVED EXAMPLE]: Missing Data Diagnostics, Imputation, & Filtering
# 1. Create an array with missing entries: data = np.array([1.0, 2.0, np.nan, 4.0, 5.0, np.nan]).
# 2. Identify missing values using np.isnan(data).
# 3. Fill NaNs with a constant (0.0) using np.nan_to_num(data, nan=0.0).
# 4. Impute missing values with the mean using np.where(np.isnan(data), np.nanmean(data), data).
# 5. Filter out NaNs completely using boolean masking: data[~np.isnan(data)].
# Hint: Standard methods (np.mean) propagate NaNs; nan-safe methods (np.nanmean) ignore them.
````
