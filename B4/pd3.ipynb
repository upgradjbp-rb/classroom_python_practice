## 🧱 Module 1: DataFrame Structural & Statistical Inspections

### 1. Concept & Theory
Before performing advanced data transformations, an analyst must inspect the structural topology and baseline distribution of the dataset:
* `df.shape`: Returns an `(n_rows, n_columns)` tuple.
* `df.columns`: Returns the index object containing column header labels.
* `df.info()`: Displays non-null counts, data types (`dtypes`), and memory utilization.
* `df.describe()`: Generates a 5-number summary (count, mean, std, min, 25%, 50%, 75%, max) for numeric features.
* **Scalar Arithmetic**: Vectorized scalar operations (e.g., `df['Col'] + 10`) broadcast across every element without explicit loops.

### 2. Syntax & Variations
```python
df.shape                     # (rows, cols)
df.columns                   # Index of column names
df.info()                    # Memory and dtype overview
df.describe()                # 8-point parametric distribution
df['Col'] + scalar_val       # Element-wise broadcasting
```

### 3. Points to Remember & Common Pitfalls
* `df.describe()` ignores non-numeric columns by default unless passed `include='all'`.
* Scalar addition `df['Col'] + 10` returns a new Series and does not mutate the DataFrame in-place unless explicitly reassigned: `df['Col'] = df['Col'] + 10`.

## ⚙️ Module 2: Custom Transformations via `.apply()` and Lambda Functions

### 1. Concept & Theory
When built-in vectorized mathematical functions do not cover custom business rules, Pandas provides the `.apply()` method. Paired with anonymous `lambda` functions, it maps custom transformations element-by-element across an existing Series to construct new features.

### 2. Syntax & Variations
```python
df['New_Col'] = df['Source_Col'].apply(lambda x: x**2)         # Mathematical exponentiation
df['Tier'] = df['Source_Col'].apply(lambda x: 'High' if x > 30 else 'Low') # Conditional mapping
```

### 3. Points to Remember & Common Pitfalls
* Chaining `.apply()` across millions of rows can be slower than native vectorized arithmetic (`df['B'] ** 2`). Use native arithmetic when available, and reserve `.apply()` for custom conditional logic.

## 📊 Module 3: Statistical Aggregations & Multi-Metric Pipelines

## 🔄 Module 4: The Split-Apply-Combine Engine (`df.groupby()`)

### 1. Concept & Theory
The **Split-Apply-Combine** paradigm handles cohort analysis:
1. **Split**: Partitions data into groups based on key columns.
2. **Apply**: Computes an aggregation (such as `.sum()` or `.mean()`) on target numerical features within each group.
3. **Combine**: Merges the aggregated outputs into a single Series or DataFrame.

### 2. Syntax & Variations
```python
df.groupby('Single_Cat')['Metric'].sum()                 # Single-key grouping
df.groupby(['Cat_A', 'Cat_B'])['Metric'].sum()           # Multi-key grouping (Hierarchical Index)
df.groupby('Single_Cat')['Metric'].sum().reset_index()   # Flatten to standard DataFrame
```

### 3. Points to Remember & Common Pitfalls
* When grouping by multiple columns `df.groupby(['ColA', 'ColB'])`, the output has a **MultiIndex**. To convert it back to a flat DataFrame with standard columns, append `.reset_index()`.

## 🔀 Module 5: Multidimensional Matrix Reshaping (`pd.pivot_table()`)

### 1. Concept & Theory
While `groupby()` produces stacked index representations, `pd.pivot_table()` reshapes data into a 2D matrix. It maps unique values of one category to the row index and another category to the column headers, aggregating numerical values at each intersection.

### 2. Syntax & Variations
```python
# Single-metric pivot table with custom aggregation (median)
pd.pivot_table(df, values='Metric', index='Row_Cat', columns='Col_Cat', aggfunc='median')

# Multi-metric pivot table with default mean aggregation
pd.pivot_table(df, values=['Metric1', 'Metric2'], index='Row_Cat', columns='Col_Cat')
```

### 3. Points to Remember & Common Pitfalls
* `aggfunc` defaults to `'mean'`. To compute sums or medians, explicitly pass `aggfunc='sum'` or `aggfunc='median'`.
* If specific index-column combinations have no underlying observations, Pandas fills those cells with `NaN` by default. You can replace them with zeros using `fill_value=0`.

## 🧱 Module 6: Frequency Contingency Matrices via Cross-Tabulation (`pd.crosstab()`)

### 1. Concept & Theory
`pd.crosstab()` computes a frequency distribution table between two or more categorical features. It counts how often specific pairs of categories co-occur, surfacing patterns without requiring manual aggregations.

### 2. Syntax & Variations
```python
pd.crosstab(df['Cat_Rows'], df['Cat_Cols'])                         # Frequency counts
pd.crosstab(df['Cat_Rows'], df['Cat_Cols'], normalize='index') * 100 # Row-wise percentages
```
