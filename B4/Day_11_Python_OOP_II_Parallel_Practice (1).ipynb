## 3. Parent + Child Constructors
A child constructor can initialize parent attributes using `super().__init__()`.

The supplied source demonstrates this parent/child initialization pattern. fileciteturn14file0

````python
class Parent:
    def __init__(self,pa):
        self.parent_attribute = pa

class Child(Parent):
    def __init__(self, pa, ca):
        super().__init__(pa) #call
        self.child_attribute = ca

child = Child("Parent Value", "Child Value")
print(child.parent_attribute)
print(child.child_attribute)
````

## 5. Method Overriding
A child class can provide its own implementation of a method inherited from its parent. This is **method overriding**. The supplied source demonstrates `ChildClass` overriding `some_method()`. fileciteturn14file0

## 8. Abstract Base Classes (ABC)
An abstract class can define behaviour that concrete child classes are expected to implement.

The supplied source uses `ABC`, `@abstractmethod`, `Shape`, `Rectangle`, and `Circle`. fileciteturn14file6

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass
````

Think of an ABC as a **contract**: a concrete Shape must provide `area()` and `perimeter()`.

````python
from abc import ABC, abstractmethod

class Shape(ABC): #abstract class
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, l, w):
        self.length = l
        self.width = w

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rectangle = Rectangle(5, 4)
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
````

## 9. Rectangle and Circle: Same Interface, Different Behaviour
The supplied source shows both subclasses implementing `area()` and `perimeter()` differently. fileciteturn14file6

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self): pass
    @abstractmethod
    def perimeter(self): pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius ** 2
    def perimeter(self):
        return 2 * 3.14 * self.radius

rectangle = Rectangle(5, 4)
circle = Circle(3)

print("Rectangle Area:", rectangle.area())
print("Rectangle Perimeter:", rectangle.perimeter())
print("Circle Area:", circle.area())
print("Circle Circumference:", circle.perimeter())
````

## 10. Operator Overloading
Python lets classes define how operators such as `+`, `-`, and `*` behave with their objects.

The supplied source calls these special methods **dunder/magic methods**. fileciteturn14file6

````python
print(5 + 10) #addition
print("Hello " + "Python") #concatenation
print([1, 2] + [3, 4]) #union/merging
# + is being overloaded
````

Different types give `+` different meanings. Operator overloading lets our own classes define their meaning too.

## 11. `__add__()` — Point Example
The supplied source uses a `Point` class and overloads `+`, `-`, and `*`. fileciteturn14file6

````python
class Point:
    def __init__(self, x, y): #magic methods
        self.x = x
        self.y = y

    def __add__(self, other): #magic methods self=p1 other=p2
        return Point(self.x + other.x, self.y + other.y) # Point(4,6) 

p1 = Point(1, 2)
p2 = Point(3, 4)

result = p1 + p2 #p1.__add__(p2)
print(result.x, result.y)
````

## 12. `__sub__()` and `__mul__()`

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Point(self.x * scalar, self.y * scalar)

p1 = Point(1, 2)
p2 = Point(3, 4)

print((p1 + p2).x, (p1 + p2).y)
print((p1 - p2).x, (p1 - p2).y)
print((p1 * 2).x, (p1 * 2).y)
````

## 15. Number Multiplication Operator Overloading
The supplied source also defines `Number.__mul__()`, checks whether the other operand is a `Number`, returns a new `Number`, and raises `TypeError` otherwise. fileciteturn14file4

````python
class Number:
    def __init__(self, value):
        self.value = value

    def __mul__(self, other):
        if isinstance(other, Number):
            return Number(self.value * other.value)
        raise TypeError("Unsupported operand type")

    def __str__(self):
        return str(self.value)

num1 = Number(5)
num2 = Number(10)
result = num1 * num2

print("Result:", result)
````

## 18. Predict Before Running

````python
class Parent:
    def show(self):
        print("Parent")

class Child(Parent):
    def show(self):
        super().show()
        print("Child")

obj = Child()
obj.show()
````

## 19. Common Beginner Mistakes

- Forgetting the parent class: `class Child(Parent):`
- Forgetting `super()` when parent initialization is required
- Trying to instantiate an abstract class with unimplemented abstract methods
- Misspelling dunder methods such as `__add__`
- Confusing **method overriding** with **operator overloading**

**Remember:** overriding changes inherited method behaviour; operator overloading defines how operators work with custom objects.
