# 4. Multiple Conditions — `and`, `or`, `not`

````python
age = 22
has_id = True
print(age >= 18 and has_id)
print(age >= 18 or has_id)
print(not has_id)
````

# 6. Why Do We Need Loops?
For thousands of transactions, we cannot write one statement per record. Loops repeat an operation efficiently.

- `while` → repeat while a condition is true
- `for` → repeat for each item in a sequence

# 7. `while` Loop

```python
while condition:
    statement
```

⚠️ Make sure the condition can eventually become false.

````python
count = 1
while count <= 5:
    print(count)
    count = count + 1
````

# 8. `break`
`break` immediately stops the loop.

````python
number = 1
while number <= 10:
    if number == 6:
        break
    print(number)
    number += 1
````

# 9. `continue`
`continue` skips the current iteration and moves to the next one.

````python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
````
