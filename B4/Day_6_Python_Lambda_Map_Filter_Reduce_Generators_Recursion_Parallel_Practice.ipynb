`sorted()` can use a lambda as its `key`.

# 4. Combining `filter()` + `map()`

A common analytics pattern is:

**Filter → Transform**

# 9. Generators

A generator produces values **one at a time** instead of creating the complete collection immediately.

The keyword `yield` turns a function into a generator.

Why useful?
- memory-efficient processing
- large datasets
- streaming-style processing
- processing values one at a time

````python
def count_up_to(n):
    current = 1

    while current <= n:
        yield current
        current += 1

for value in count_up_to(5):
    print(value)
````

## Generator for Data Processing

Instead of storing every transformed result in a list, a generator can produce each result when requested.

````python
def squared_values(numbers):
    for number in numbers:
        yield number ** 2

data = [2, 4, 6, 8]

for value in squared_values(data):
    print(value)
````

# 10. Recursion

A recursive function calls itself.

A recursive function needs:
1. **Base case** — when to stop.
2. **Recursive case** — how the function moves toward the base case.

Think of recursion as solving a smaller version of the same problem.

````python
def countdown(n):
    if n == 0:
        print("Done!")
        return

    print(n)
    countdown(n - 1)

countdown(5)
````

# 11. Source-Based Recursive Example — Equal Values

The source material also contains an `equal()` style problem. The important learning objective here is understanding how a recursive solution can compare or process values repeatedly.

Use the practice cell to experiment with recursive comparisons.

````python
def equal(a, b):
    if len(a) != len(b):
        return False

    if len(a) == 0:
        return True

    if a[0] != b[0]:
        return False

    return equal(a[1:], b[1:])

print(equal([1, 2, 3], [1, 2, 3]))
print(equal([1, 2, 3], [1, 4, 3]))
````
