## 1. Understand the types of data

**Numerical data** describes quantities that can be counted or measured, such as runs scored, sales, age, price, or marks.

- **Discrete numerical data:** values usually counted in whole numbers, such as number of orders.

**Categorical data** describes groups or labels, such as product category, city, department, or player name.

**Why this matters:** chart selection depends on the question and the data types. A line plot often shows change over time; a scatter plot explores the relationship between two numerical variables; a bar chart compares categories; a histogram shows the distribution of one numerical variable; a pie chart shows parts of a whole.

## 2. Line plots

### What is a line plot?
A line plot connects data points in order. It is especially useful for showing how a numerical value changes across time or another ordered variable.

**Real-world examples**
- Monthly sales across a year
- Daily temperature
- A player's runs across seasons
- Website visitors week by week

### Key idea
A line chart is usually most meaningful when the x-axis has a natural order. It can mislead when unrelated categories are connected as if they form a continuous sequence.

**Typical syntax to learn:** `plot(x_values, y_values)`  
The x-values determine horizontal positions; y-values determine vertical positions.

### Titles, axis labels, colors, line styles, and markers

A chart should answer three questions without requiring the instructor to explain it verbally:
1. What is being compared?
2. What does the horizontal axis mean?
3. What does the vertical axis mean?

- **X label / Y label:** describes the meaning and units of each axis.
- **Color:** helps distinguish series; do not rely on color alone to communicate meaning.
- **Line style:** solid, dashed, dotted, or dash-dot.
- **Line width:** controls line thickness.
- **Marker:** symbol drawn at each data point, such as a circle or diamond.
- **Marker size:** controls the size of those symbols.

## 3. Scatter plots

### What is a scatter plot?
A scatter plot displays individual observations as dots. It is useful when comparing **two numerical variables** and exploring whether they appear related.

**Real-world examples**
- Advertising spend versus sales
- Study hours versus exam marks
- Batting average versus strike rate
- Product price versus customer rating

Each dot represents one observation. A general upward pattern may suggest a positive relationship; a downward pattern may suggest a negative relationship. A cloud without a clear pattern may suggest little obvious relationship.

**Caution:** a visible relationship does not prove that one variable causes the other.

## 4. Bar charts

### What is a bar chart?
A bar chart compares values across categories. Bar length or height represents the numerical value.

**Examples:** sales by product, learners by course, runs by player, or expenses by category.

- Use a **vertical bar chart** when category labels are short or there are few categories.
- Use a **horizontal bar chart** when labels are long.
- For ordinary bar charts, start the numerical axis at zero when possible because bar length encodes magnitude.

Bar charts are commonly used for grouped summaries, not to show the distribution of every individual numerical observation.

## 5. Histograms

### What is a histogram?
A histogram groups numerical values into intervals called **bins** and counts how many observations fall into each interval.

**Examples:** distribution of student marks, order values, delivery times, or runs scored in innings.

A histogram is for one numerical variable. Unlike a bar chart, its bars represent numerical intervals, so the order of bins matters and bars usually touch.

### Important vocabulary
- **Bin:** an interval or range.
- **Bin edges:** the boundaries that define intervals.
- **Frequency:** number of observations in an interval.
- **Distribution:** the overall pattern of values, including concentration, spread, skew, and possible unusual values.

## 6. Pie charts

### What is a pie chart?
A pie chart displays how categories contribute to a whole. Each slice represents a share of the total.

**Use it when:** there are a small number of categories, values are non-negative, categories form a meaningful whole, and the differences are easy to see.

**Avoid it when:** there are too many categories, values are negative, categories overlap, or precise comparisons matter. A sorted bar chart is often easier to compare.

- **Labels:** identify each slice.
- **Percentage labels:** show each slice's share of the total.
- **Colors:** distinguish categories.
- **Explode:** separates a slice visually from the rest.
- **Shadow:** adds a decorative shadow; use it only if it improves clarity.
