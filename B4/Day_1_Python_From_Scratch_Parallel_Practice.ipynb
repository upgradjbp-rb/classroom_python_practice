# 3. Variables — Python's labelled data boxes



# 7. Arithmetic Operators

Python can work as a calculator.

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition | `10 + 5` |

# 11. String Slicing — extracting a portion

Indexing gives **one character**.

Slicing gives **multiple characters**.

### Syntax

```python
string[start:stop]
```

The `start` position is included.

The `stop` position is excluded.

For:

```python
word = "PYTHON"
```

```python
word[0:2]   # PY
word[2:5]   # THO
```

### Full slicing syntax

```python
string[start:stop:step]
```

Examples:

```python
word[:3]     # first 3 characters
word[3:]     # from index 3 to end
word[:]      # complete string
word[::2]    # every second character
word[::-1]   # reverse
```

````python
word = "PYTHON"

print(word[0:2])
print(word[2:5])
print(word[:3])
print(word[3:])
print(word[::2])
print(word[::-1])
````

### The most important slicing rule

Remember:

> **Start is included. Stop is excluded.**

Think of `0:3` as:

> Start at 0, stop **before** 3.

This same idea will become extremely useful later with lists, arrays and data.

# 13. `split()` and `join()` — moving between text and pieces

Suppose a customer record is stored as:

```text
"Neha,Bhopal,Electronics,85000"
```

We can split it:

```python
record.split(",")
```

Result:

```python
['Neha', 'Bhopal', 'Electronics', '85000']
```

We can join a list back into text:

```python
",".join(parts)
```

This is extremely useful when working with text-based data.

````python
record = "Neha,Bhopal,Electronics,85000"

parts = record.split(",")

print(parts)
print("Customer:", parts[0])
print("City:", parts[1])
print("Category:", parts[2])
print("Sales:", parts[3])
````
