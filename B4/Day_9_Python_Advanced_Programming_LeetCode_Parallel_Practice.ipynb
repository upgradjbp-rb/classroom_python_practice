# 12. Subarray Sum Equals K

Count how many subarrays have a sum equal to `k`.

Example:
```text
nums = [1, 1, 1]
k = 2
Output → 2
```

### Source approach
Use:
- a running prefix sum
- a dictionary counting previous prefix sums

If the current prefix sum is `cur`, we need a previous prefix sum equal to `cur - k`.

````python
# Instructor Example — Subarray Sum Equals K
from collections import defaultdict

def subarray_sum(nums, k):
    count = defaultdict(int)
    count[0] = 1
    current = 0
    result = 0

    for x in nums:
        current += x
        result += count[current - k]
        count[current] += 1

    return result

print(subarray_sum([1, 1, 1], 2))
````

### Complexity
- Time: **O(n)**
- Space: **O(n)**
