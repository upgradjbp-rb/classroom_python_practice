# 1. Why do we need Data Structures?

Suppose a store has one sale:

```python
sale = 2500
```

Easy.

But what if it has 10,000 sales?

Creating thousands of separate variables would be painful:

```python
sale1 = 2500
sale2 = 1800
sale3 = 4200
...
```

Instead, we can group related data:

```python
sales = [2500, 1800, 4200, ...]
```

This is the central idea of **data structures**.

### Think like an analyst

| Requirement | Useful structure |
|---|---|
| Ordered collection of sales | List |
| Fixed collection of coordinates/categories | Tuple |
| Customer → total sales | Dictionary |
| Unique cities/categories | Set |

# 6. Adding Elements — `append()`, `insert()`, `extend()`

### `append()`

Adds **one item at the end**.

```python
sales.append(9000)
```

### `insert()`

Adds an item at a specific position.

```python
sales.insert(1, 9000)
```

### `extend()`

Adds multiple items from another collection.

```python
sales.extend([9000, 10000])
```

### Remember

- `append()` → one item at the end
- `insert()` → one item at a chosen position
- `extend()` → multiple items

````python
sales = [5000, 7000, 8000]

sales.append(9000)
print("After append :", sales)

sales.insert(1, 6000)
print("After insert :", sales)

sales.extend([10000, 11000])
print("After extend :", sales)
````

# 7. Removing Elements — `remove()`, `pop()`, `del`

### `remove(value)`

Removes the first matching value.

```python
products.remove("Mouse")
```

### `pop(index)`

Removes and **returns** an item.

```python
removed = products.pop(1)
```

If no index is supplied, it removes the last item.

### `del`

Deletes using an index or slice.

```python
del products[1]
```

### ⚠️ Difference

`remove()` → "remove this value"

`pop()` → "remove this position and give me the removed item"

`del` → delete by position/slice

````python
products = ["Laptop", "Mouse", "Keyboard", "Monitor"]

products.remove("Mouse")
print(products)

removed = products.pop(1)
print("Removed:", removed)
print(products)

del products[0]
print(products)
````

# 9. Tuples — ordered but fixed

A tuple is an **ordered collection that cannot be changed after creation**.

### Syntax

```python
my_tuple = (item1, item2, item3)
```

Example:

```python
coordinates = (23.18, 79.94)
```

### Properties
- ordered;
- indexed;
- allows duplicates;
- immutable;
- useful when data should remain fixed.

### Real-world analogy

A person's date of birth is not something your program should casually modify.

A coordinate pair or fixed configuration can naturally be represented as a tuple.

````python
coordinates = (23.18, 79.94)

print(coordinates)
print(coordinates[0])
print(coordinates[-1])
print(len(coordinates))
````

### Tuple methods

Because tuples are immutable, they have fewer methods.

Commonly used:

```python
count()
index()
```

````python
ratings = (5, 4, 5, 3, 5, 4)

print(ratings.count(5))
print(ratings.index(3))
````

### ⚠️ List vs Tuple

```python
sales = [100, 200, 300]      # can change
sales_fixed = (100, 200, 300) # cannot change
```

Trying:

```python
sales_fixed[0] = 500
```

causes an error.

This distinction becomes important when designing reliable programs.

# 10. Dictionaries — data with labels

A dictionary stores data as **key-value pairs**.

Think of it like a form:

```text
Name       → Neha
City       → Bhopal
Sales      → 85000
```

### Syntax

```python
my_dict = {
    "key1": value1,
    "key2": value2
}
```

Example:

```python
customer = {
    "name": "Neha",
    "city": "Bhopal",
    "sales": 85000
}
```

### Why dictionaries are powerful in analytics

Instead of remembering:

```python
customer[0]
customer[1]
customer[2]
```

we can write:

```python
customer["name"]
customer["sales"]
```

The label explains what the value means.

````python
customer = {
    "name": "Neha",
    "city": "Bhopal",
    "sales": 85000
}

print(customer)
print(customer["name"])
print(customer["sales"])
````

### Accessing safely with `get()`

```python
customer.get("sales")
```

If the key does not exist, `get()` can return `None` instead of immediately raising a key error.

You can also provide a default:

```python
customer.get("age", "Not Available")
```

````python
customer = {"name": "Neha", "sales": 85000}

print(customer.get("sales"))
print(customer.get("age", "Not Available"))
````

# 13. Sets — the unique-values machine

A set is a collection designed to contain **unique values**.

### Syntax

```python
my_set = {item1, item2, item3}
```

Example:

```python
cities = {"Bhopal", "Indore", "Bhopal", "Jabalpur"}
```

Duplicate `"Bhopal"` is retained only once.

### Why analysts care

Real datasets often contain repeated categories.

If we need to answer:

> "Which unique cities appear in this data?"

a set is a natural tool.

````python
cities = {"Bhopal", "Indore", "Bhopal", "Jabalpur", "Indore"}

print(cities)
print(len(cities))
````

### Important set properties

- unique elements;
- unordered;
- no indexing like lists;
- mutable as a collection;
- excellent for membership and set operations.

⚠️ Do not rely on a set's displayed order.

### Creating an empty set

This is a common beginner trap:

```python
{}
```

creates an empty dictionary.

Use:

```python
set()
```

for an empty set.

````python
empty_dictionary = {}
empty_set = set()

print(type(empty_dictionary))
print(type(empty_set))
````

# 14. Adding and removing set elements

### Add

```python
cities.add("Ujjain")
```

### Remove

```python
cities.remove("Indore")
```

### Safer removal

```python
cities.discard("Indore")
```

`discard()` does not raise an error if the item is absent.

````python
cities = {"Bhopal", "Indore", "Jabalpur"}

cities.add("Ujjain")
print(cities)

cities.discard("Indore")
print(cities)
````

# 15. Set Operations

Sets become particularly useful when comparing groups.

### Union — everything from both

```python
A | B
```

or:

```python
A.union(B)
```

### Intersection — common values

```python
A & B
```

or:

```python
A.intersection(B)
```

### Difference — values only in A

```python
A - B
```

### Symmetric difference — values in either set, but not both

```python
A ^ B
```

````python
online_customers = {"Aarav", "Neha", "Riya", "Kabir"}
store_customers = {"Neha", "Kabir", "Simran", "Tanya"}

print("All customers:", online_customers | store_customers)
print("Both channels:", online_customers & store_customers)
print("Only online:", online_customers - store_customers)
print("Only one channel:", online_customers ^ store_customers)
````

# 16. Choosing the Right Data Structure

This is more important than memorizing methods.

| Question | Choose |
|---|---|
| Do I need an ordered, changeable collection? | List |
| Do I need fixed ordered data? | Tuple |
| Do I need label → value mapping? | Dictionary |
| Do I need unique values? | Set |

### Example

A sales analyst has:

```text
All monthly sales → List
Company coordinates → Tuple
Customer → Total sales → Dictionary
Unique cities → Set
```

There is no single "best" data structure. The right choice depends on the job.
