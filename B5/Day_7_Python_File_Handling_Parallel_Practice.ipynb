# 9. File Data + String Processing

Suppose each line contains `Name,Marks`.

Workflow:
1. Read the line
2. `strip()` the newline
3. `split(",")`
4. Convert marks to `int`
5. Process the data

````python
with open("marks.txt", "r") as file:
    for line in file:
        name, marks = line.strip().split(",")
        marks = int(marks)
        print(name, marks)
````

# 11. Creating a Simple Report File

Real-world pattern:

**Input file → Process data → Output report**

````python
total = 0
count = 0

with open("marks.txt", "r") as file:
    for line in file:
        name, marks = line.strip().split(",")
        total += int(marks)
        count += 1

average = total / count

with open("marks_report.txt", "w") as report:
    report.write("Student Marks Report\n")
    report.write("--------------------\n")
    report.write(f"Number of students: {count}\n")
    report.write(f"Average marks: {average:.2f}\n")

print("Report generated.")
````
