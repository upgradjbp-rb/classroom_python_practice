## 9. Why do we need `__init__()`?

Manually assigning attributes becomes repetitive when many objects are required.

Python provides the special `__init__()` method, which is automatically called when a new instance is created and is commonly used to initialize its attributes. fileciteturn12file0

````python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

student1 = Student("Aman", 20, "Data Science")

print(student1.name)
print(student1.age)
print(student1.course)
````

### Read this carefully

`self.name = name`

means:

> Take the incoming `name` value and store it as this object's `name` attribute.

`self` is the current object; `name` is the incoming parameter.

## 13. A method can change object state

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def show_balance(self):
        print(self.owner, "has balance:", self.balance)

account = BankAccount("Aman", 5000)

account.show_balance()
account.deposit(2000)
account.show_balance()
````

## 17. Complete beginner OOP example

Let's combine:
- class
- constructor
- attributes
- methods
- multiple objects

````python
class Student:
    def __init__(self, name, age, course, marks):
        self.name = name
        self.age = age
        self.course = course
        self.marks = marks

    def show_details(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
        print("Marks:", self.marks)

    def is_passed(self):
        return self.marks >= 40

student1 = Student("Aman", 20, "Data Science", 78)
student2 = Student("Priya", 21, "AI", 35)

student1.show_details()
print("Passed:", student1.is_passed())

print()

student2.show_details()
print("Passed:", student2.is_passed())
````

## 18. Predict before running

Trace the values of `p1` and `p2` before executing.

````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def discount_price(self, discount):
        return self.price - (self.price * discount / 100)

p1 = Product("Laptop", 50000)
p2 = Product("Phone", 30000)

print(p1.discount_price(10))
print(p2.discount_price(20))
````

## 19. Identify the OOP parts

Identify the class, constructor, attributes, method and objects.

````python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department

    def show(self):
        print(self.name, "-", self.department)

e1 = Employee("Neha", "Analytics")
e2 = Employee("Ravi", "HR")

e1.show()
e2.show()
````

### Answer
- **Class:** `Employee`
- **Constructor:** `__init__`
- **Attributes:** `name`, `department`
- **Method:** `show`
- **Objects:** `e1`, `e2`

## 20. Debugging exercise

Find the problem before running this code.

````python
class Student:
    def __init__(self, name):
        self.name = name #problem

    def show(self):
        print(self.name)

student1 = Student("Aman")
student1.show() #amanm
student1.name = "Arjun"
student1.show() #arjun
````

### Fix

The object attribute must be assigned using:

`self.name = name`

````python
class Student:
    def __init__(self, name):
        self.name = name

    def show(self):
        print(self.name)

student1 = Student("Aman")
student1.show()
````

## 21. Common beginner mistakes

### Mistake 1 — forgetting `self`
Instance methods normally use `self` as their first parameter.

### Mistake 2 — writing `name = name`
Usually you want `self.name = name`.

### Mistake 3 — confusing class and object
`Student` is the class; `student1` is an object.

### Mistake 4 — calling an instance method without an object
Normally use `student1.show()`.
