# 1. What Is an Algorithm?

An **algorithm** is a sequence of steps used to solve a problem.

### Example: Find the largest of three numbers

Instead of immediately writing code:

1. Take three numbers.
2. Assume the first is largest.
3. Compare the second with the current largest.
4. Compare the third with the current largest.
5. Display the largest.

### Golden Rule

**Do not start with Python syntax. Start with the logic.**

## Algorithm Thinking Template

For every problem, ask:

1. **Input:** What information do I receive?
2. **Process:** What calculations or decisions are required?
3. **Output:** What should I display/return?
4. **Repetition:** Do I need a loop?
5. **Decision:** Do I need a condition?
6. **Storage:** Do I need a variable, list, counter or accumulator?

# 7. Problem Decomposition — Digits of a Number

A number such as:

`123456`

can be processed one digit at a time.

Two useful operations:

- `num % 10` → extracts the last digit
- `num // 10` → removes the last digit

Example:

```text
123456 % 10 → 6
123456 // 10 → 12345
```

Repeat this until the number becomes `0`.

## Mini Practice

What digits will be printed for `5072`?

Run the program after predicting.

````python
num = 5072

while num > 0:
    digit = num % 10
    print(digit)
    num //= 10
````

# 12. Common Beginner Algorithm Mistakes

### Mistake 1 — Wrong loop boundary

`range(1, n)` does **not** include `n`.

### Mistake 2 — Forgetting to update

A `while` loop can become infinite.

### Mistake 3 — Initialising the accumulator incorrectly

For a sum, usually start with `0`.

### Mistake 4 — Updating the wrong variable

When processing digits, remember:

```python
digit = num % 10
num //= 10
```

### Mistake 5 — Coding before understanding

Always write the steps first.
