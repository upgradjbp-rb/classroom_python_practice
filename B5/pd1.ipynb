## 🧱 Part 1: Pandas Series — Foundations & Constructors

### 1. Concept & Theory
A **Pandas Series** is a one-dimensional labeled array capable of holding any data type (integers, floats, strings, Python objects). The axis labels are collectively referred to as the **index**. A Series can be constructed from Python lists, NumPy arrays, or Python dictionaries.

### 2. Syntax & Variations
```python
pd.Series(data=my_list)                     # Default integer index (0, 1, 2, ...)
pd.Series(data=my_list, index=labels)       # Explicit custom labels
pd.Series(data=numpy_array)                 # From NumPy ndarray
pd.Series(data=dictionary)                  # Keys automatically become index labels
```

### 3. Points to Remember & Common Pitfalls
* When creating a Series from a dictionary, the dictionary keys automatically become the index, and the dictionary values become the data values.
* If you specify custom `index` labels alongside a list or array, the length of the labels must match the length of the data exactly; otherwise, a `ValueError: Length of values does not match length of index` will be raised.

## 📊 Part 2: DataFrame Construction — Dictionaries vs. 2D Lists

### 1. Concept & Theory
A **DataFrame** is a 2D labeled data structure with columns of potentially different data types. It can be thought of as a spreadsheet or SQL table. It can be constructed:
1. From a **dictionary of lists**: Keys become column names; lists become column values.
2. From a **list of lists (2D array)**: Requires an explicit `columns` list parameter to assign meaningful header names.

### 2. Syntax & Variations
```python
# Constructor 1: Dictionary of lists
df = pd.DataFrame(data_dict)

# Constructor 2: List of row lists
df = pd.DataFrame(data_list, columns=['Col1', 'Col2', 'Col3'])
```

### 3. Points to Remember & Common Pitfalls
* In a dictionary constructor, all lists must be of identical length; otherwise, Python raises `ValueError: All arrays must be of the same length`.
* If you omit the `columns` argument when passing a 2D list, Pandas assigns integer column headers (`0, 1, 2, ...`).

## 📑 Part 3: Column Selection, Mutation, & Removal

### 1. Concept & Theory
Columns in a DataFrame are manipulated as follows:
* **Single Column Selection**: `df['ColName']` returns a 1D `Series`.
* **Multiple Column Selection**: `df[['ColA', 'ColB']]` (note the double brackets) returns a 2D `DataFrame` subset.
* **Adding Columns**: `df['NewCol'] = [values]` assigns new column data.
* **Removing Rows/Columns**: `df.drop(label, axis=0)` drops rows; `df.drop(columns=['ColName'])` or `df.drop('ColName', axis=1)` drops columns.

### 2. Syntax & Variations
```python
# Column selection
subset = df[['Col1', 'Col2']]

# New column assignment
df['NewCol'] = ['A', 'B', 'C', 'D']

# Dropping a row by index label
df.drop(0, axis=0)                          # Returns copy unless inplace=True

# Dropping a column
df.drop('ColName', axis=1)
```

### 3. Points to Remember & Common Pitfalls
* **The Drop Immutability Rule**: `.drop()` does not alter the DataFrame in-place by default; it returns a new modified copy. You must either reassign `df = df.drop(...)` or pass `inplace=True`.
* **Axis Mechanics**: `axis=0` refers to rows (horizontal drop), while `axis=1` refers to columns (vertical drop).

## 🎯 Part 4: Row Selection & Subsetting (`loc` vs. `iloc`)

### 1. Concept & Theory
Pandas provides two primary methods for slicing rows and sub-matrices:
* `df.loc[]`: **Label-based** indexing. Uses index labels and column header strings.
* `df.iloc[]`: **Integer-position based** indexing. Uses 0-indexed integer coordinates ($0$ to $N-1$).

### 2. Syntax & Variations
```python
# Selecting multiple rows by index labels
df.loc[[0, 1]]

# Selecting a single row by 0-based integer position
df.iloc[3]

# Selecting a sub-matrix (rows + columns)
df.loc[[0, 1], ['ColA', 'ColB']]
df.loc[[0, 1]][['ColA', 'ColB']]
```

### 3. Points to Remember & Common Pitfalls
* `df.iloc[3]` retrieves the row at the 4th numerical position (0, 1, 2, 3), regardless of what index label that row has.
* In label slicing with `.loc['a':'c']`, the endpoint `'c'` is **inclusive**. In integer slicing with `.iloc[0:2]`, the stop position `2` is **exclusive**.

## 🔍 Part 5: Conditional Filtering & Boolean Slicing

### 1. Concept & Theory
Filtering rows based on business logic is accomplished through **Boolean indexing**:
1. An evaluation expression `df['Col'] > threshold` generates a boolean Series of `True` and `False` values.
2. Passing this boolean Series back into the DataFrame `df[boolean_mask]` extracts only the matching records.
3. Multiple conditions must use bitwise operators (`&` for AND, `|` for OR), and each sub-condition must be enclosed in parentheses `(cond1) & (cond2)`.

### 2. Syntax & Variations
```python
# Single condition filter
df[df['NumericCol'] > 30]

# Compound condition filter (AND)
df[(df['ColA'] > 30) & (df['ColB'] == 'TargetValue')]

# Compound condition filter (OR)
df[(df['ColA'] > 30) | (df['ColB'] == 'TargetValue')]
```

### 3. Points to Remember & Common Pitfalls
* **The Parentheses Requirement**: In Python, bitwise operators (`&`, `|`) have higher precedence than comparison operators (`>`, `<`). Writing `df[df['Age'] > 30 & df['City'] == 'Paris']` without parentheses will result in an `Ambiguous truth value` error.
* **String Equality**: String comparisons inside conditional masks are case-sensitive by default (`'Paris' != 'paris'`).
