## 🧩 Part 1: Relational Merging (`pd.merge()`)

### 1. Concept & Theory
`pd.merge()` connects rows in DataFrames based on one or more shared keys, directly mirroring relational database SQL `JOIN` operations:
* **`how='inner'`**: Retains only records where the key exists in **both** tables (intersection).
* **`how='outer'`**: Retains all records from both tables, populating missing matches with `NaN` (union).
* **`how='left'`**: Preserves every record from the left table, joining matching right-table attributes.
* **`how='right'`**: Preserves every record from the right table, joining matching left-table attributes.

### 2. Syntax & Variations
```python
pd.merge(left_df, right_df, on='key_column', how='inner')
pd.merge(left_df, right_df, on='key_column', how='outer')
pd.merge(left_df, right_df, on='key_column', how='left')
pd.merge(left_df, right_df, on='key_column', how='right')
```

### 3. Points to Remember & Common Pitfalls
* **Type Alignment**: The key column must share the exact same data type in both tables (e.g., merging an integer ID with a string ID yields an empty DataFrame or `MergeError`).
* **Non-Matching Rows**: Outer, left, and right merges introduce `NaN` values for non-matching records, which automatically upcasts integer columns to `float64`.

## 🗂️ Part 2: DataFrame Concatenation (`pd.concat()`)

### 1. Concept & Theory
`pd.concat()` appends or stacks DataFrames along a designated axis:
* **`axis=0` (Row Stacking / Default)**: Stacks DataFrames vertically. Ideal when combining identical batches of data collected over multiple days or regions.
* **`axis=1` (Column Side-by-Side Binding)**: Binds DataFrames horizontally side-by-side along matching index coordinates.

### 2. Syntax & Variations
```python
pd.concat([df1, df2])                   # Vertical concatenation (axis=0)
pd.concat([df1, df2], axis=1)            # Horizontal concatenation (axis=1)
pd.concat([df1, df2], ignore_index=True) # Resets continuous integer index
```

### 3. Points to Remember & Common Pitfalls
* When concatenating along `axis=0`, identical index labels from both tables are preserved by default (`0, 1, 2, 0, 1, 2`). To produce a continuous unique index, pass `ignore_index=True`.
* When concatenating along `axis=1`, DataFrames are aligned by their row index labels, not row positions.

## 🔗 Part 3: Index-Based Joining (`.join()`)

### 1. Concept & Theory
While `pd.merge()` connects tables via column values, `.join()` connects DataFrames primarily using their **index labels**:
* By default, `df1.join(df2)` performs a **left join** on the index of `df1`.
* Setting `how='outer'` preserves all unique index values from both tables.

### 2. Syntax & Variations
```python
df1.join(df2)              # Default left join on matching index
df1.join(df2, how='outer') # Outer join on index
df1.join(df2, how='inner') # Inner join on index
```

### 3. Points to Remember & Common Pitfalls
* `.join()` requires that the column names in both tables be unique unless explicit prefixes or suffixes are supplied (`lsuffix='_left'`, `rsuffix='_right'`); otherwise, a `ValueError: columns overlap` occurs.

## 🔍 Part 4: Missing Data Diagnostics (`isna()`, `isna().sum()`, `isna().any()`)

### 1. Concept & Theory
Missing values in Pandas are denoted as `np.nan` (Not a Number) or `None`:
* `df.isna()` (or `df.isnull()`): Produces a boolean DataFrame of identical dimensions (`True` where missing, `False` where valid).
* `df.isna().sum()`: Aggregates missing values column-wise, reporting the total missing count per feature.
* `df.isna().any()`: Returns a boolean Series indicating whether each column contains at least one missing entry.

### 2. Syntax & Variations
```python
df.isna()         # Boolean matrix
df.isna().sum()   # Total null count per column
df.isna().any()   # Boolean check per column
```

## 🗑️ Part 5: Removing Missing Data (`dropna()` & `thresh`)

### 1. Concept & Theory
When imputing values is statistically inappropriate, rows or columns with missing data must be removed:
* `df.dropna()`: Drops any row containing **at least one** `NaN` entry.
* `df.dropna(thresh=k)`: Retains rows that have at least `k` valid, non-null values (threshold filtering).
* `df.dropna(axis=1)`: Drops entire columns containing `NaN` values.

### 2. Syntax & Variations
```python
df.dropna()                    # Drops rows with any missing value
df.dropna(thresh=1)            # Keeps rows with >= 1 non-null value
df.dropna(thresh=3)            # Keeps rows with >= 3 non-null values
df.dropna(subset=['ColA'])     # Evaluates nullity only on specified columns
```

### 3. Points to Remember & Common Pitfalls
* Calling `dropna()` indiscriminately can delete a large portion of your data if missing values are scattered across different rows.
* Like `.drop()`, `.dropna()` returns a copy by default unless reassigned or passed `inplace=True`.

## 🩹 Part 6: Missing Data Imputation (`fillna()`)

### 1. Concept & Theory
When dropping rows is unacceptable, missing values must be imputed:
* **Constant Scalar Imputation**: `df.fillna(0)` replaces all missing cells with a fixed constant.
* **Dictionary-Based Custom Imputation**: `df.fillna(value={'ColA': valA, 'ColB': valB})` applies specific domain defaults per column.
* **Statistical Summary Imputation**: `df.fillna(df.mean())` replaces missing cells with the column's mean value.

### 2. Syntax & Variations
```python
df.fillna(0)                             # Fills all NaNs with 0
df.fillna(value={'ColA': 0, 'ColB': 100}) # Specific column imputation
df.fillna(df.mean())                     # Dynamic column mean imputation
```

### 3. Points to Remember & Common Pitfalls
* `df.mean()` only calculates means on numeric columns. Running `df.fillna(df.mean())` on a DataFrame that contains string columns will raise a `TypeError` in modern Pandas versions unless you explicitly select numeric columns: `df.fillna(df.mean(numeric_only=True))`.

````python
# [SOLVED EXAMPLE]: Imputing Missing Values with Scalars, Dictionaries, and Column Means[cite: 7]
# What to write:[cite: 7]
# 1. Constant scalar fill: fill_zero = survey_df.fillna(0)[cite: 7]
# 2. Column-specific dictionary map: fill_custom = survey_df.fillna(value={'A': 0.0, 'B': 100.0, 'C': 300.0, 'D': 400.0})[cite: 7]
# 3. Statistical mean fill: fill_mean = survey_df.fillna(survey_df.mean())[cite: 7]
# Hint: survey_df.mean() dynamically calculates column averages (ignoring NaNs) and fills respective features in a single step[cite: 7]
````
