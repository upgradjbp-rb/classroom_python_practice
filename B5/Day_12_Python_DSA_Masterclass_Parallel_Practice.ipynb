## 1. What is DSA?

**Data Structure** = a way of organizing/storing data.

Examples: List, Dictionary, Set, Tuple.

**Algorithm** = a step-by-step method to solve a problem.

**Analogy:** A supermarket's shelves/categories are the data structure; the method used to find an item is the algorithm.

## 2. Why DSA Matters in Data Science

With 10 records, almost any approach feels fast. With millions of records, efficiency matters.

Common Data Science connections:
- frequency counting
- customer/product lookup
- duplicate detection
- running totals
- searching sorted data
- rolling time windows
- efficient array processing

## 3. Big-O — Basic Idea

| Complexity | Simple interpretation |
|---|---|
| O(1) | Almost constant work |
| O(log n) | Grows very slowly |
| O(n) | Roughly one pass through input |
| O(n log n) | Common efficient sorting pattern |
| O(n²) | Nested-loop style growth |

**Memory trick:** O(1) = one action; O(n) = everything once; O(n²) = many things compared with many others.

## Frequency Counting

A common pattern in analytics is counting how often each item occurs.

````python
products = ["Laptop", "Phone", "Laptop", "Tablet", "Phone", "Laptop"]
frequency = {}
for product in products:
    frequency[product] = frequency.get(product, 0) + 1
print(frequency)
````

# Pattern 2 — Two Sum

Given numbers and a target, find two numbers whose sum equals the target.

Example: `[2, 7, 11, 15]`, target `9` → `2 + 7`.

**Key idea:** for `x`, the required partner is `target - x`. A dictionary remembers values already seen.

````python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        needed = target - num
        if needed in seen:
            return [seen[needed], i]
        seen[num] = i
    return []

print(two_sum([2, 7, 11, 15], 9))
````

# Pattern 3 — Majority Element

A majority element appears more than `n/2` times.

Example: `[2,2,1,1,1,2,2]` → `2`.

### Boyer–Moore Voting Idea
Same candidate → increase support.
Different candidate → cancel support.
When a majority is guaranteed, the remaining candidate is the answer.

````python
def majority_element(nums):
    candidate = None
    count = 0
    for num in nums:
        if count == 0:
            candidate = num
        if num == candidate:
            count += 1
        else:
            count -= 1
    return candidate

print(majority_element([2, 2, 1, 1, 1, 2, 2]))
````

# Pattern 4 — Prefix Sum

A prefix sum stores cumulative totals.

`[100, 200, 150, 300]` → `[100, 300, 450, 750]`.

It is useful for repeated range-sum questions and cumulative analytics.

````python
sales = [100, 200, 150, 300]
prefix = []
running_total = 0
for value in sales:
    running_total += value
    prefix.append(running_total)
print(prefix)
````

### Data Analytics Connection
A cumulative-sales column answers: **How much have we sold up to this date?**

# Pattern 5 — Subarray Sum Equals K

Given an array and `k`, count continuous subarrays whose sum equals `k`.

Example: `[1,1,1]`, `k=2` → `2`.

**Main idea:** running prefix sum + dictionary of previously seen prefix sums.

````python
def subarray_sum(nums, k):
    prefix_count = {0: 1}
    prefix_sum = 0
    count = 0
    for num in nums:
        prefix_sum += num
        count += prefix_count.get(prefix_sum - k, 0)
        prefix_count[prefix_sum] = prefix_count.get(prefix_sum, 0) + 1
    return count

print(subarray_sum([1, 1, 1], 2))
````

### Think
Why do we look for `prefix_sum - k`? Think about the difference between two cumulative totals.

# Pattern 6 — Sliding Window

A sliding window processes a continuous portion of data while moving the window forward.

Examples:
- last 7 days of sales
- rolling average
- moving activity count
- longest/shortest substring satisfying a condition

**Analogy:** You move a train window forward rather than rebuilding the whole view.

````python
numbers = [2, 1, 5, 1, 3, 2]
window_size = 3
window_sum = sum(numbers[:window_size])
maximum = window_sum

for right in range(window_size, len(numbers)):
    window_sum += numbers[right]
    window_sum -= numbers[right - window_size]
    maximum = max(maximum, window_sum)

print("Maximum window sum:", maximum)
````

# Pattern 10 — Binary Search

Binary search works on **sorted data**:
1. inspect the middle
2. eliminate half
3. repeat

**Analogy:** finding a word in a dictionary by opening near the middle.

````python
def binary_search(nums, target):
    left = 0
    right = len(nums) - 1
    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

numbers = [10, 20, 30, 40, 50, 60]
print(binary_search(numbers, 40))
````

**Memory trick:** Before Binary, check the **Order**.

# Mini Challenge

Daily sales: `[120, 80, 150, 200, 90, 170, 210]`

Find the maximum total for any **3 consecutive days**.

Which pattern fits?

A. Binary Search  
B. Sliding Window  
C. Majority Vote  
D. Two Sum

**Answer: B — Sliding Window.**

````python
sales = [120, 80, 150, 200, 90, 170, 210]
window_size = 3
window_sum = sum(sales[:window_size])
maximum = window_sum
for right in range(window_size, len(sales)):
    window_sum += sales[right]
    window_sum -= sales[right - window_size]
    maximum = max(maximum, window_sum)
print(maximum)
````
