### Why Seaborn?

- provides a layer of abstraction hence simpler to use
- better aesthetics
- more graphs included

**Simple analogy:** Matplotlib gives you the basic drawing tools; Seaborn adds convenient defaults and statistical plotting helpers so you can make informative charts with less setup.

### Seaborn Roadmap

Types of Functions

- Figure Level
- Axis Level

Main Classification

- Relational Plot
- Distribution Plot
- Categorical Plot
- Regression Plot
- Matrix Plot
- Multiplots

https://seaborn.pydata.org/api.html

**Figure-level functions** manage the whole figure and often support facets. **Axes-level functions** draw on a particular set of axes and are useful for fine-grained control.
