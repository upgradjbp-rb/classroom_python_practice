## 9. `try` + `except` + `else`

The `else` block executes only when the `try` block completes without an exception.

```python
# Instructor Example
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Invalid number.")
else:
    print("Conversion successful.")
    print("Number:", number)
```

## 10. `finally`

The `finally` block runs whether an exception occurs or not. It is useful for cleanup work.

```python
# Instructor Example
try:
    number = int(input("Enter a number: "))
    print(number)
except ValueError:
    print("Invalid input.")
finally:
    print("Program execution finished.")
```

## 12. Invalid Data from a File

```python
# Instructor Example
data = ["85", "91", "seventy", "76"]

for value in data:
    try:
        marks = int(value)
        print("Valid marks:", marks)
    except ValueError:
        print("Invalid marks found:", value)
```

## 14. Debugging Mindset

When Python shows an exception, do not panic.

1. Read the exception name.
2. Read the last line of the traceback.
3. Find the line number.
4. Understand which operation failed.
5. Check the values and their types.
6. Fix the cause rather than hiding the problem.
7. Run the program again.

**Example:** If Python reports `ZeroDivisionError`, ask: *Where am I dividing, and can the denominator become zero?*
