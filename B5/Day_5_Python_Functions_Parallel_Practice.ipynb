# Common Beginner Mistakes

### 1. Forgetting the function call
Defining a function does not automatically execute it.

### 2. Confusing `print` and `return`
`print()` displays. `return` sends a value back.

### 3. Wrong number of arguments
Match the required parameters unless defaults are provided.

### 4. Returning too early
Statements after an executed `return` do not run.

### 5. Scope confusion
Variables created inside functions are normally local to that function.
